```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/Mall_Customers.csv")
```

```python
df.head()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Age"], bins=10)

plt.title("Age Distribution of Customers")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.show()
```

## Insight
## - The histogram shows how customers are distributed across different age groups.
## - Most customers belong to the age groups with the tallest bars.
## - This helps identify the primary age segment of the mall's customers.

plt.figure(figsize=(8,5))

plt.hist(df["Annual Income (k$)"], bins=10)

plt.title("Annual Income Distribution")
plt.xlabel("Annual Income (k$)")
plt.ylabel("Number of Customers")

plt.show()

## Insight

## - The histogram shows how customers are distributed based on their annual income.
## - It helps identify whether most customers belong to low-, middle-, or high-income groups.

```python
plt.figure(figsize=(8,5))

plt.hist(df["Spending Score (1-100)"], bins=10)

plt.title("Spending Score Distribution")
plt.xlabel("Spending Score")
plt.ylabel("Number of Customers")

plt.show()
```

## Insight

## - The histogram displays the distribution of customer spending scores.
## - It helps identify whether customers generally have low, medium, or high spending behavior.

# Bivariate Analysis

## In this section, we analyze the relationship between two variables to identify trends and patterns.

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Age"], df["Spending Score (1-100)"])

plt.title("Age vs Spending Score")
plt.xlabel("Age")
plt.ylabel("Spending Score")

plt.show()
```

### Observation

## - This scatter plot shows the relationship between customers' age and spending score.
## - It helps identify whether spending habits vary with age.
## - The points indicate that customers of different age groups have different spending behaviors.

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Annual Income (k$)"], df["Spending Score (1-100)"])

plt.title("Annual Income vs Spending Score")
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score")

plt.show()
```

### Observation

## - This scatter plot compares customers' annual income with their spending score.
## - It helps identify whether higher income is associated with higher spending.
## - Different customer groups can be observed based on income and spending behavior.

```python
plt.figure(figsize=(8,5))

df.boxplot(column="Annual Income (k$)", by="Gender")

plt.title("Annual Income by Gender")
plt.suptitle("")
plt.xlabel("Gender")
plt.ylabel("Annual Income (k$)")

plt.show()
```

### Observation

## - The box plot compares the annual income distribution of male and female customers.
## - It shows the median, spread, and possible outliers for each gender.

```python
correlation = df.corr(numeric_only=True)
correlation
```

```python
plt.figure(figsize=(8,6))

plt.imshow(correlation, cmap="coolwarm", interpolation="nearest")
plt.colorbar()

plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=45)
plt.yticks(range(len(correlation.columns)), correlation.columns)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```

### Observation

## - The heatmap shows the correlation between numerical features.
## - Correlation values close to **1** indicate a strong positive relationship.
## - Values close to **-1** indicate a strong negative relationship.
## - Values close to **0** indicate little or no linear relationship.
